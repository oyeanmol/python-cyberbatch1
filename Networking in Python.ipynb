# Networking with Python

Python provides built-in support for network programming through modules such as socket. Before working with sockets, it is important to understand IP addresses, ports, protocols, clients, servers, TCP, and UDP.

## 1. What is Networking?

Computer networking is the process of connecting computers or devices so that they can communicate and exchange data.

For example:

```
Your Computer  ───────────>  Web Server
     │                           │
     │       Request             │
     ├──────────────────────────>│
     │                           │
     │       Response            │
     │<──────────────────────────┤

```
When you open:

```https://google.com```

your computer communicates with Google's servers through a network.

Python can also communicate with other computers using networking.

## 2. IP Address

An IP address identifies a device on a network.

Think of an IP address like the address of a house.

If you want to send something to a particular house, you need its address.

Similarly, if one computer wants to communicate with another computer, it needs its IP address.

```Example
192.168.1.10
```

This is an example of an IPv4 address.

### IPv4

IPv4 addresses contain four numbers separated by dots.

```192.168.1.10```

Each number can range from:
```
0 - 255
```
Example:
```
10.0.0.1
172.16.0.5
192.168.1.100
8.8.8.8
```

### IPv6

IPv6 is the newer IP addressing system.

Example:
```
2001:4860:4860::8888
```
IPv6 was introduced because the number of available IPv4 addresses is limited.

## 3. Public and Private IP Addresses
Private IP

Private IP addresses are generally used inside local networks.

Common private ranges include:
```
10.0.0.0 - 10.255.255.255
172.16.0.0 - 172.31.255.255
192.168.0.0 - 192.168.255.255
```
Example:
```
192.168.1.20
```
Your laptop might have a private IP like this when connected to your Wi-Fi.

Public IP

A public IP identifies your network/device to the public Internet.

Example:
```
203.0.113.25
```
The actual public IP assigned to you depends on your ISP/network configuration.

## 4. localhost

localhost refers to your own computer.

The commonly used IPv4 localhost address is:
```
127.0.0.1
```
For example:
```
127.0.0.1:5000
```
means:

Connect to port 5000 on this same computer.

You will frequently use localhost while developing and testing networking applications.

## 5. Port

An IP address identifies the computer.

A port identifies a particular network service/application on that computer.

Think about it like this:
```
IP Address = Building
Port       = Room
```
For example:
```
192.168.1.10:80
```
means:
```
Computer: 192.168.1.10
Port:     80
```
Why do we need ports?

A computer can run many network applications simultaneously.

For example:
```
Web Server       → Port 80
HTTPS Server     → Port 443
SSH Server       → Port 22
Application      → Port 5000
```
The IP tells the data which computer to reach.

The port tells the computer which application should receive the data.

## 6. Port Numbers

Port numbers range from:
```
0 - 65535
```
They are commonly divided into:
```
0–1023 — Well-known ports
```
Examples:
```bash
Port	Common Service
20/21	FTP
22	SSH
23	Telnet
25	SMTP
53	DNS
80	HTTP
443	HTTPS
1024–49151
```
Registered ports commonly used by applications.
```
49152–65535
```
Dynamic/private ports commonly used for temporary connections.

The exact usage can vary by operating system and application.

## 7. Protocol

A protocol is a set of rules that defines how devices communicate.

Think of it as a language or communication rule.

For example:
```
Human → English/Hindi
Computer → Network Protocol
```

Different networking tasks use different protocols.

## 8. TCP

### TCP = Transmission Control Protocol

TCP provides a reliable, connection-oriented communication mechanism.

Important characteristics:
```
Connection-oriented
Reliable delivery
Data arrives in order
Lost data can be retransmitted
Uses acknowledgements and other control mechanisms
```
Example:
```
Client                    Server

   SYN  ------------------->
        <------------------- SYN + ACK
   ACK  ------------------->
   
       Connection Established

After the connection is established, data can be exchanged.
```
TCP Example
```
Imagine transferring:

Hello World

TCP attempts to ensure that the receiver gets the data reliably and in the correct order.

TCP is commonly used for:

HTTP/HTTPS
SSH
FTP
Email protocols
```
## 9. UDP

UDP = User Datagram Protocol

UDP is connectionless and has much less protocol overhead than TCP.

Characteristics:

#### Connectionless
 - No built-in guarantee of delivery
 - No guarantee of ordering
 - Lower overhead
 - Often useful where speed and low latency matter

Examples include applications such as:

 - DNS
 - Online gaming
 - Streaming/real-time communication

A simplified comparison:

 - TCP	UDP
 - Connection-oriented	Connectionless
 - Reliable delivery mechanism	No delivery guarantee
 - Ordered byte stream	Individual datagrams
 - More overhead	Less overhead
 - Common for web/file transfer	Common for real-time traffic
## 10. Client and Server

Most network applications use the concept of a client and server.

Server

The server waits for incoming connections.

Client

The client connects to the server.

Example:
```
             Server
        192.168.1.10:5000
               ↑
               │
         Connection
               │
               ↓
             Client
        192.168.1.20
```
A simple web example:
```
Browser = Client
Web Server = Server
```
