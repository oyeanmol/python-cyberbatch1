```python
import re

successful_login = 0
failed_login = 0
log_file = "security.log"

with open(log_file, "r") as f:
    for line in f:
        ip = re.search(r"ip=(\d+\.\d+\.\d+\.\d+)", line)
        user = re.search(r"user=(\w+)", line)
        if "Failed login" in line:
            failed_login += 1
            print(f"[Failed] User: {user} | IP: {ip}")
        elif "successful" in line:
            successful_login += 1
            print(f"[Success] User: {user} | IP: {ip}")
print("\n----- Security Log Summary-----")
print("Successful Logins: ", successful_login)
print("Failed Logins: ", failed_login)
```

```python
import re

file = "ssh.log"
ip_count = {}

with open(file, "r") as f:
    for line in f:
        if "Failed password" in line:
            match = re.search(r"from (\d+\.\d+\.\d+\.\d+)", line)

            if match:
                ip = match.group(1)

                if ip in ip_count:
                    ip_count[ip] += 1
                else:
                    ip_count[ip] = 1

print("\n----- Failed Login Attempts -------")
for ip, count in ip_count.items():
    print(ip, "-> ", count, "Attemps")
print("\n---- Suspicious IPs -------")
for ip, count in ip_count.items():
    if count > 2:
        print(ip, "-> ", count, "failed attempts")
```
