```python
default = "SecurePassword123"
count = 0

while True:
    user = input("Enter your password")
    count += 1
    if user == default:
        print("Welcome Home")
        break
    else:
        print("Try Again")
        if count == 4:
            print("\n\nThis your Last Attempt!")
        if count == 5:
            print("You have Attempted maximum time. Please try Again in 30 seconds.")
            break
    
```

```python
blacklist = ['192.168.1.10', '10.0.0.5', '172.16.0.1', '203.0.113.45']
observed = ['192.168.1.1', '10.0.0.5', '192.168.1.100', '203.0.113.45', '198.51.100.22']


for ip in observed:
    if ip in blacklist:
        print("\n[Alert]This is malicious IP: ", ip)
    else:
        print("\n[info] This is clean IP: ", ip)
```
