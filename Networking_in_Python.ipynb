# Networking with Python

Python provides built-in support for network programming through modules such as `socket`. Before working with sockets, it is important to understand **IP addresses, ports, protocols, clients, servers, TCP, and UDP**.

---


## 1. What is Networking?

**Computer networking** is the process of connecting computers or devices so that they can communicate and exchange data.

For example:

```text
Your Computer  ───────────>  Web Server
     │                           │
     │       Request             │
     ├──────────────────────────>│
     │                           │
     │       Response            │
     │<──────────────────────────┤
```

When you open:

```text
https://google.com
```

your computer communicates with Google's servers through a network.

Python can also communicate with other computers using networking.

# 2. IP Address

An **IP address** identifies a device on a network.

Think of an IP address like the **address of a house**.

If you want to send something to a particular house, you need its address.

Similarly, if one computer wants to communicate with another computer, it needs its IP address.

### Example

```text
192.168.1.10
```

This is an example of an IPv4 address.

---


## IPv4

IPv4 addresses contain four numbers separated by dots.

```text
192.168.1.10
```

Each number can range from:

```text
0 - 255
```

Example:

```text
10.0.0.1
172.16.0.5
192.168.1.100
8.8.8.8
```

---


## IPv6

IPv6 is the newer IP addressing system.

Example:

```text
2001:4860:4860::8888
```

IPv6 was introduced because the number of available IPv4 addresses is limited.

# 5. Port

An IP address identifies the **computer**.

A port identifies a particular **network service/application** on that computer.

Think about it like this:

```text
IP Address = Building
Port       = Room
```

For example:

```text
192.168.1.10:80
```

means:

```text
Computer: 192.168.1.10
Port:     80
```

---


## Why do we need ports?

A computer can run many network applications simultaneously.

For example:

```text
Web Server       → Port 80
HTTPS Server     → Port 443
SSH Server       → Port 22
Application      → Port 5000
```

The IP tells the data **which computer to reach**.

The port tells the computer **which application should receive the data**.

# 8. TCP

**TCP = Transmission Control Protocol**

TCP provides a reliable, connection-oriented communication mechanism.

Important characteristics:

* Connection-oriented
* Reliable delivery
* Data arrives in order
* Lost data can be retransmitted
* Uses acknowledgements and other control mechanisms

Example:

```text
Client                    Server

   SYN  ------------------->
        <------------------- SYN + ACK
   ACK  ------------------->
   
       Connection Established
```

After the connection is established, data can be exchanged.

---


## TCP Example

Imagine transferring:

```text
Hello World
```

TCP attempts to ensure that the receiver gets the data reliably and in the correct order.

TCP is commonly used for:

```text
HTTP/HTTPS
SSH
FTP
Email protocols
```

---


# 9. UDP

**UDP = User Datagram Protocol**

UDP is connectionless and has much less protocol overhead than TCP.

Characteristics:

* Connectionless
* No built-in guarantee of delivery
* No guarantee of ordering
* Lower overhead
* Often useful where speed and low latency matter

Examples include applications such as:

```text
DNS
Online gaming
Streaming/real-time communication
```

A simplified comparison:

| TCP                          | UDP                          |
| ---------------------------- | ---------------------------- |
| Connection-oriented          | Connectionless               |
| Reliable delivery mechanism  | No delivery guarantee        |
| Ordered byte stream          | Individual datagrams         |
| More overhead                | Less overhead                |
| Common for web/file transfer | Common for real-time traffic |

# 27. Mini Lab 1 — TCP Port Scanner

A port scanner checks whether a TCP service is accepting connections on a given port.

### Important

Only scan systems you own or have explicit permission to test.

For learning, use:

```text
127.0.0.1
```

---


## Simple Port Scanner

```python
import socket

host = "127.0.0.1"

for port in range(1, 1025):

    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

    sock.settimeout(0.5)

    result = sock.connect_ex((host, port))

    if result == 0:
        print(f"Port {port} is OPEN")

    sock.close()
```

---


## How it works

Suppose:

```text
host = 127.0.0.1
```

The program tries:

```text
127.0.0.1:1
127.0.0.1:2
127.0.0.1:3
...
127.0.0.1:1024
```

For each port:

```python
connect_ex()
```

attempts a TCP connection.

If the return value is:

```python
0
```

the connection succeeded.

Therefore:

```python
if result == 0:
    print("OPEN")
```

---


# 28. Better Port Scanner

We can make the program reusable:

```python
import socket

def scan_port(host, port):

    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    sock.settimeout(0.5)

    result = sock.connect_ex((host, port))

    sock.close()

    return result == 0


host = "127.0.0.1"

for port in range(1, 1025):

    if scan_port(host, port):
        print(f"Port {port} is OPEN")
```

# 29. Mini Lab 2 — TCP Chat Application

Now we'll build a simple two-way chat.

We will create:

```text
server.py
client.py
```

---


## Server

```python
import socket

server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

server.bind(("127.0.0.1", 5000))

server.listen(1)

print("Waiting for client...")

client, address = server.accept()

print("Connected:", address)

while True:

    message = client.recv(1024).decode()

    if message.lower() == "exit":
        print("Client disconnected")
        break

    print("Client:", message)

    reply = input("You: ")

    client.sendall(reply.encode())

    if reply.lower() == "exit":
        break

client.close()
server.close()
```

---


## Client

```python
import socket

client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

client.connect(("127.0.0.1", 5000))

while True:

    message = input("You: ")

    client.sendall(message.encode())

    if message.lower() == "exit":
        break

    reply = client.recv(1024).decode()

    print("Server:", reply)

    if reply.lower() == "exit":
        break

client.close()
```

# 32. Mini Lab 3 — File Transfer

Now we'll create a basic TCP file-transfer program.

The client sends a file to the server.

We'll keep the example simple and use a small header containing the filename.

---


## Server — `file_server.py`

```python
import socket

server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

server.bind(("127.0.0.1", 5000))
server.listen(1)

print("Waiting for client...")

client, address = server.accept()

print("Connected:", address)

filename = client.recv(1024).decode()

with open("received_" + filename, "wb") as file:

    while True:

        data = client.recv(4096)

        if not data:
            break

        file.write(data)

print("File received successfully")

client.close()
server.close()
```

---


# 33. Client — `file_client.py`

```python
import socket
import os

filename = "example.txt"

client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

client.connect(("127.0.0.1", 5000))

client.sendall(os.path.basename(filename).encode())

with open(filename, "rb") as file:

    while True:

        data = file.read(4096)

        if not data:
            break

        client.sendall(data)

print("File sent successfully")

client.close()
```

---


# 34. How File Transfer Works

Suppose the client has:

```text
example.txt
```

The client opens it in binary mode:

```python
open(filename, "rb")
```

Then reads:

```text
4096 bytes
```

at a time.

```text
File
 │
 ├── 4096 bytes ──────>
 ├── 4096 bytes ──────>
 ├── 4096 bytes ──────>
 └── remaining bytes ─>
```

The server receives the bytes:

```python
data = client.recv(4096)
```

and writes them:

```python
file.write(data)
```

until:

```python
recv()
```

returns:

```text
b''
```

which indicates that the peer has performed an orderly shutdown and there is no more data to read.

# 37. TCP Is a Stream

This is one of the most important socket concepts.

Suppose you send:

```python
sock.sendall(b"Hello")
sock.sendall(b"World")
```

The receiver is **not guaranteed** to receive:

```text
Hello
World
```

as two separate `recv()` calls.

It might receive:

```text
HelloWorld
```

or:

```text
Hel
loworld
```

or another grouping.

Therefore, applications need a method of **message framing**.

Common approaches include:

```text
Fixed-size messages
Length prefix
Delimiter
Structured protocol
Connection close
```

---


# 38. Example of Length-Prefixed Message

Suppose we want to send:

```text
Hello
```

We can first send its length.

Conceptually:

```text
0005Hello
```

The receiver first reads the length:

```text
0005
```

Then knows that it needs to read:

```text
5 bytes
```

This approach is much more reliable than assuming one `recv()` equals one message.

---


# 39. `recv()` Does NOT Mean "Receive One Message"

This misconception is very common.

When you write:

```python
data = sock.recv(1024)
```

it means approximately:

> Give me up to 1024 bytes currently available from this TCP stream.

It does **not** mean:

> Give me exactly one message.

This distinction becomes extremely important when building production networking applications.

# 42. Handling Socket Errors

Network operations can fail.

For example:

```python
import socket

try:

    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

    sock.settimeout(2)

    sock.connect(("127.0.0.1", 5000))

    print("Connected")

except socket.timeout:
    print("Connection timed out")

except ConnectionRefusedError:
    print("Connection refused")

except OSError as e:
    print("Network error:", e)

finally:
    sock.close()
```

# 44. Simple UDP Example

### UDP Server

```python
import socket

server = socket.socket(
    socket.AF_INET,
    socket.SOCK_DGRAM
)

server.bind(("127.0.0.1", 5000))

print("Waiting for message...")

data, address = server.recvfrom(1024)

print("Client:", data.decode())

server.sendto(
    b"Hello from UDP server!",
    address
)

server.close()
```

### UDP Client

```python
import socket

client = socket.socket(
    socket.AF_INET,
    socket.SOCK_DGRAM
)

client.sendto(
    b"Hello UDP server!",
    ("127.0.0.1", 5000)
)

data, address = client.recvfrom(1024)

print("Server:", data.decode())

client.close()
```

Notice that UDP does not require the same connection establishment process as TCP.

# 46. Multi-Client Server Using Threads

Once the basic concept is understood, a simple threaded server can handle multiple clients.

```python
import socket
import threading


def handle_client(client, address):

    print("Connected:", address)

    while True:

        data = client.recv(1024)

        if not data:
            break

        print(address, ":", data.decode())

        client.sendall(data)

    client.close()


server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

server.bind(("127.0.0.1", 5000))
server.listen()

print("Server started...")

while True:

    client, address = server.accept()

    thread = threading.Thread(
        target=handle_client,
        args=(client, address)
    )

    thread.start()
```

This introduces an important real-world concept:

```text
Main Server
     │
     ├── Client 1 → Thread 1
     ├── Client 2 → Thread 2
     ├── Client 3 → Thread 3
     └── Client 4 → Thread 4
```

## Final Concept to Remember

The most important thing students should take away is this:

```text
                 NETWORKING
                     │
             IP + PORT + PROTOCOL
                     │
                  SOCKET
                     │
          ┌──────────┴──────────┐
          │                     │
        SERVER                CLIENT
          │                     │
       socket()              socket()
          │                     │
       bind()                connect()
          │                     │
      listen()                  │
          │                     │
      accept()                  │
          └──────────┬──────────┘
                     │
                send / recv
                     │
              DATA TRANSFER
                     │
          ┌──────────┼──────────┐
          │          │          │
        Scanner     Chat       File
                              Transfer
```

**Core idea:** an **IP address identifies where**, a **port identifies which service**, a **protocol defines how communication works**, and a **socket gives your Python program the interface through which that communication happens**.
