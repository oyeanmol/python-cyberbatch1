# OS & Environment Interaction with Python

This notebook provides a detailed overview and examples for interacting with the operating system and environment using Python's built-in modules like `os`, `sys`, and `subprocess`. We will also cover path handling and environment variables, concluding with two mini-labs.

## 1. The `os` and `sys` Modules

### The `os` Module

The `os` module provides a way of using operating system dependent functionality. It allows you to interact with the operating system in various ways, such as creating directories, changing the current working directory, listing directory contents, and more.

os module as a bridge between your Python program and your computer's operating system.

It lets Python do things like:

Create folders
Delete folders/files
Find out where your program is running
See files inside a folder
Rename files
Work with paths
Read environment variables
Run operating-system commands

````python
import os

# Get the current working directory
current_directory = os.getcwd()
print(f"Current Working Directory: {current_direc tory}")

# List contents of the current directory
print("\nContents of current directory:")
for item in os.listdir(current_directory):
    print(f"- {item}")

# Create a new directory
new_dir = 'my_new_directory'
if not os.path.exists(new_dir):
    os.mkdir(new_dir)
    print(f"\nCreated directory: {new_dir}")
else:
    print(f"\nDirectory '{new_dir}' already exists.")

# Change the current working directory
os.chdir(new_dir)
print(f"Changed to new working directory: {os.getcwd()}")

# Go back to the original directory (assuming it was the parent)
os.chdir('..')
print(f"Changed back to original working directory: {os.getcwd()}")

# Remove the directory (it must be empty)
if os.path.exists(new_dir) and not os.listdir(new_dir):
    os.rmdir(new_dir)
    print(f"Removed directory: {new_dir}")
elif os.path.exists(new_dir):
    print(f"Directory '{new_dir}' is not empty or could not be removed.")
````

### The `sys` Module

The `sys` module provides access to system-specific parameters and functions. It allows interaction with the interpreter itself, such as retrieving Python version information, managing search paths, and exiting the program.

````python
import sys

# Get Python version information
print(f"Python Version: {sys.version}")

# Get the platform (e.g., 'linux', 'win32', 'darwin')
print(f"Operating System Platform: {sys.platform}")

# Get the list of directories where Python looks for modules
print("\nPython Module Search Path (sys.path):")
for path in sys.path:
    print(f"- {path}")
````

## 2. Path Handling

The `os.path` submodule is crucial for manipulating file and directory paths in an OS-agnostic way. It provides functions to join, split, check existence, and get components of paths.

````python
import os

# Example Path
path = "/home/user/documents/report.txt"

# Join path components
directory = "/home/user/documents"
filename = "report.txt"
full_path = os.path.join(directory, filename)
print(f"Joined path: {full_path}")

# Check if a path exists
print(f"Does '{path}' exist? {os.path.exists(path)}")

# Check if it's a file or directory
# (Note: These will return False for a path that doesn't actually exist)
print(f"Is '{path}' a file? {os.path.isfile(path)}")
print(f"Is '{directory}' a directory? {os.path.isdir(directory)}")

# Get directory name and base name
dir_name = os.path.dirname(path)
base_name = os.path.basename(path)
print(f"Directory Name: {dir_name}")
print(f"Base Name: {base_name}")

# Split path into head and tail
head, tail = os.path.split(path)
print(f"Split path (head): {head}")
print(f"Split path (tail): {tail}")

# Get file extension and base name without extension
file_name, file_extension = os.path.splitext(path)
print(f"File name without extension: {file_name}")
print(f"File extension: {file_extension}")
````

## 3. Environment Variables

Environment variables are dynamic named values that can affect the way running processes will behave. They are part of the operating system environment and can be accessed by programs. Python's `os` module provides access to these variables via `os.environ`.

````python
import os

# Access an environment variable (e.g., PATH, HOME, USER)
# Note: Not all variables might be set in all environments.

# Example 1: PATH variable
path_var = os.environ.get('PATH')
print(f"PATH environment variable: {path_var[:100]}...") # Print only first 100 chars for brevity

# Example 2: HOME or USER variable
home_var = os.environ.get('HOME') # Unix-like systems
user_var = os.environ.get('USER') # Unix-like systems
username_var = os.environ.get('USERNAME') # Windows systems

print(f"\nHOME environment variable: {home_var}")
print(f"USER environment variable: {user_var}")
print(f"USERNAME environment variable: {username_var}")

# Set a new environment variable (temporary for the current process)
os.environ['MY_TEST_VAR'] = 'Hello Colab!'
print(f"\nMY_TEST_VAR (after setting): {os.environ.get('MY_TEST_VAR')}")

# Delete an environment variable
del os.environ['MY_TEST_VAR']
print(f"MY_TEST_VAR (after deleting): {os.environ.get('MY_TEST_VAR')}")
````

## 4. Running Shell Commands with `subprocess`

The `subprocess` module allows you to spawn new processes, connect to their input/output/error pipes, and obtain their return codes. It's the recommended way to run shell commands from Python.

2. The most basic example

On Windows:

```python
import subprocess
```

```python
subprocess.run(["dir"], shell=True)
```

On Linux/macOS:

```python
import subprocess

subprocess.run(["ls"])
```

The command gets executed by your operating system.

3. Why not just use os.system()?

You might remember:

```python
os.system("dir")
```

That can also execute a command.

But subprocess is generally much more powerful.

With subprocess, you can:

 - Run a command
 - Capture its output
 - Capture errors
 - Send input to the program
 - Check whether it succeeded
 - Control the process
 - Set environment variables
 - Connect multiple processes together

So:
```
os.system()
     ↓
"Run this command"

subprocess
     ↓
"Run this command AND let me control what happens"
```

````python
import subprocess

# Example 1: Run a simple command and capture output
print("--- Running 'ls -l' ---")
result = subprocess.run(['ls', '-l'], capture_output=True, text=True, check=True)
print(result.stdout)

# Example 2: Run a command with arguments
print("--- Running 'echo Hello World!' ---")
result = subprocess.run(['echo', 'Hello World!'], capture_output=True, text=True, check=True)
print(result.stdout)

# Example 3: Running a command directly in the shell (use with caution due to shell injection risks)
# This is less safe if parts of the command come from user input.
print("--- Running 'pwd' (via shell=True) ---")
result = subprocess.run('pwd', shell=True, capture_output=True, text=True, check=True)
print(result.stdout)

# Example 4: Handling errors
print("--- Running a non-existent command (expecting error) ---")
try:
    subprocess.run(['non_existent_command'], capture_output=True, text=True, check=True)
except subprocess.CalledProcessError as e:
    print(f"Error: {e.stderr.strip()}")
except FileNotFoundError as e:
    print(f"Error: {e}")
````

## 5. Mini Lab: Script to find `.exe` files in a given directory

This script will recursively search a specified directory for all files ending with `.exe` (executables on Windows systems). For demonstration purposes in a Colab environment (which is Linux-based), we will simulate `.txt` files as our target.

````python
import os

def find_files_with_extension(directory, extension):
    """
    Recursively finds all files with a given extension in a directory.
    """
    found_files = []
    for root, _, files in os.walk(directory):
        for file in files:
            if file.endswith(extension):
                found_files.append(os.path.join(root, file))
    return found_files

# Create some dummy files for demonstration
dummy_dir = 'test_search_dir'
os.makedirs(os.path.join(dummy_dir, 'subdir1'), exist_ok=True)
os.makedirs(os.path.join(dummy_dir, 'subdir2'), exist_ok=True)

with open(os.path.join(dummy_dir, 'file1.txt'), 'w') as f: f.write('content')
with open(os.path.join(dummy_dir, 'program.exe'), 'w') as f: f.write('binary') # Simulating .exe
with open(os.path.join(dummy_dir, 'subdir1', 'another.txt'), 'w') as f: f.write('content')
with open(os.path.join(dummy_dir, 'subdir2', 'app.exe'), 'w') as f: f.write('binary') # Simulating .exe
with open(os.path.join(dummy_dir, 'document.pdf'), 'w') as f: f.write('pdf content')

# Define the directory to search and the extension
search_directory = dummy_dir
extension_to_find = '.exe'

print(f"Searching for '{extension_to_find}' files in '{search_directory}'...")
exe_files = find_files_with_extension(search_directory, extension_to_find)

if exe_files:
    print("Found files:")
    for f in exe_files:
        print(f"- {f}")
else:
    print(f"No '{extension_to_find}' files found.")

# Clean up dummy directory and files
import shutil
shutil.rmtree(dummy_dir)
print(f"\nCleaned up '{dummy_dir}'.")
````

## Discovering `os` module methods

To see all the functions and attributes available in a module like `os`, you can use the built-in `dir()` function. It returns a list of names in the current scope or a list of attributes for an object.

````python
import os

# Get all attributes and methods of the os module
os_members = dir(os)

print("All members of the 'os' module:")
for member in os_members:
    print(f"- {member}")

# If you want to see only callable methods (functions), you can filter them:
print("\nCallable methods of the 'os' module:")
for member in os_members:
    if callable(getattr(os, member)):
        print(f"- {member}")
````
