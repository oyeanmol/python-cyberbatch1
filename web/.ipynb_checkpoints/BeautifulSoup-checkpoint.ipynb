## 1. The `requests` Module

The `requests` library is an elegant and simple HTTP library for Python, designed to make HTTP requests simpler and more human-friendly. It allows you to send HTTP requests and handle their responses, making it the foundation for web scraping, API interaction, and more.

````python
# First, install the requests library if you haven't already
# !pip install requests

import requests

# Make a simple GET request to a website
url = 'https://www.example.com'
response = requests.get(url)

# Print the status code (200 means success)
print(f"Status Code: {response.status_code}")

# Print the first 500 characters of the response content (HTML)
print("\nResponse Content (first 500 chars):\n", response.text[:500])
````

## 2. The `BeautifulSoup` Module

`BeautifulSoup` is a Python library for parsing HTML and XML documents. It creates a parse tree from page source code that can be used to extract data in a hierarchical and readable way. It's excellent for navigating, searching, and modifying parse trees.

````python
# First, install the beautifulsoup4 library if you haven't already
# !pip install beautifulsoup4

from bs4 import BeautifulSoup

# Let's use the HTML content from our previous requests example
html_doc = requests.get('https://www.example.com').text

# Parse the HTML content
soup = BeautifulSoup(html_doc, 'html.parser')

# Example 1: Get the title of the page
print(f"Page Title: {soup.title.string}")

# Example 2: Find the first <h1> tag
h1_tag = soup.find('h1')
if h1_tag:
    print(f"\nFirst H1 tag text: {h1_tag.text}")

# Example 3: Find all <p> tags
all_p_tags = soup.find_all('p')
print(f"\nNumber of P tags: {len(all_p_tags)}")
if all_p_tags:
    print(f"First P tag text: {all_p_tags[0].text}")

# Example 4: Find an element by ID (if 'my_id' exists)
# For example, if there was an element <div id="my_id">...</div>
# my_div = soup.find(id='my_id')
# if my_div:
#     print(f"Content of div with ID 'my_id': {my_div.text}")
````

## 3. Combining `requests` and `BeautifulSoup` (A Simple Example)

This shows a typical workflow: fetch HTML with `requests`, then parse and extract data with `BeautifulSoup`.

````python
import requests
from bs4 import BeautifulSoup

url = 'https://www.python.org/' # Python's official website
response = requests.get(url)
response.raise_for_status() # Raise an exception for HTTP errors

soup = BeautifulSoup(response.text, 'html.parser')

# Extract the main navigation links
print("Main Navigation Links:")
nav_links = soup.select('.main-navigation ul.menu li a')
for link in nav_links:
    print(f"- {link.text.strip()} ({link.get('href')})")
````

## 5. Explanation of Key `requests` and `BeautifulSoup` Functions/Methods

Let's break down the important functions and methods used in the previous examples from both libraries.

### `requests` Module Functions and Attributes

1.  **`requests.get(url, **kwargs)`**
    *   **Purpose:** Sends an HTTP GET request to the specified URL.
    *   **Example Usage:**
        ```python
        import requests
        url = 'https://www.example.com'
        # Send a GET request with a timeout of 5 seconds
        response = requests.get(url, timeout=5,
                                 headers={'User-Agent': 'MyCustomBrowser'})
        print(f"Request to {url} sent.")
        ```
    *   **Explanation:** This is the primary way to fetch content from a web page. It returns a `Response` object. `kwargs` like `timeout` (in seconds) can be used to control the request behavior, and `headers` can be used to send custom HTTP headers (e.g., `User-Agent`).

2.  **`response.status_code`**
    *   **Purpose:** An integer attribute of the `Response` object that indicates the HTTP status code of the response.
    *   **Example Usage:**
        ```python
        # Assuming 'response' is from a previous requests.get()
        if response.status_code == 200:
            print('Request successful (Status: 200 OK)!')
        elif response.status_code == 404:
            print('Page not found (Status: 404)!')
        else:
            print(f'Request failed with status code: {response.status_code}')
        ```
    *   **Explanation:** `200` means "OK" (request succeeded), `404` means "Not Found", `500` means "Internal Server Error", etc. It's crucial for checking if your request was successful.

3.  **`response.text`**
    *   **Purpose:** A string attribute of the `Response` object that contains the content of the response, decoded by `requests` based on the response headers.
    *   **Example Usage:**
        ```python
        # Assuming 'response' is from a previous requests.get()
        html_content = response.text
        print('First 100 characters of HTML content:')
        print(html_content[:100])
        ```
    *   **Explanation:** For web scraping, this usually contains the HTML source code of the web page. This is what you'll pass to `BeautifulSoup`.

4.  **`response.raise_for_status()`**
    *   **Purpose:** Checks if an HTTP request returned a successful status code. If not, it raises an `HTTPError`.
    *   **Example Usage:**
        ```python
        # Assuming 'response' is from a previous requests.get()
        try:
            response.raise_for_status() # This will raise an HTTPError for 4xx/5xx responses
            print('Request was successful (no HTTP error).')
        except requests.exceptions.HTTPError as e:
            print(f'HTTP Error occurred: {e}')
        ```
    *   **Explanation:** This is a convenient way to immediately stop execution if a request failed (e.g., due to a 404 or 500 error), preventing further processing of potentially invalid or empty content.

5.  **`requests.exceptions` (HTTPError, ConnectionError, Timeout, RequestException)**
    *   **Purpose:** These are various exception types that `requests` can raise when something goes wrong during an HTTP request.
    *   **Example Usage:**
        ```python
        import requests
        try:
            # Example: Trying to connect to a non-existent domain to trigger ConnectionError
            # or a very long timeout to trigger Timeout
            response = requests.get('http://nonexistent-domain-123456.com', timeout=1)
            response.raise_for_status()
        except requests.exceptions.ConnectionError as e:
            print(f"Connection Error: {e}")
        except requests.exceptions.Timeout as e:
            print(f"Timeout Error: {e}")
        except requests.exceptions.HTTPError as e:
            print(f"HTTP Error: {e}")
        except requests.exceptions.RequestException as e:
            # This will catch any other requests-related exception
            print(f"An unexpected requests error occurred: {e}")
        ```
    *   **Explanation:**
        *   `HTTPError`: Raised by `response.raise_for_status()` for bad HTTP responses.
        *   `ConnectionError`: Raised for network-related errors (e.g., DNS failure, refused connection).
        *   `Timeout`: Raised if the request exceeds the specified `timeout`.
        *   `RequestException`: The base exception for all `requests` exceptions. Catching this will catch any of the above and more, providing a general error handling mechanism.

### `BeautifulSoup` Module Functions and Attributes

1.  **`BeautifulSoup(markup, parser)`**
    *   **Purpose:** The constructor for creating a `BeautifulSoup` object, which represents the parsed HTML or XML document.
    *   **Example Usage:**
        ```python
        from bs4 import BeautifulSoup
        html_doc = """<html><head><title>My Page</title></head>
                    <body><h1>Welcome</h1><p>Hello, world!</p></body></html>"""
        soup = BeautifulSoup(html_doc, 'html.parser')
        print(f"BeautifulSoup object created successfully.")
        ```
    *   **Explanation:** `markup` is the raw HTML/XML string (e.g., from `response.text`). `parser` specifies which parser to use; `'html.parser'` is a built-in Python parser, but `lxml` or `html5lib` can also be used if installed for better performance or handling of malformed HTML.

2.  **`soup.title.string`**
    *   **Purpose:** An attribute that directly accesses the text content of the `<title>` tag within the parsed HTML.
    *   **Example Usage:**
        ```python
        # Assuming 'soup' is from the previous BeautifulSoup example
        if soup.title:
            page_title = soup.title.string
            print(f"Page Title: {page_title}")
        else:
            print("No title tag found.")
        ```
    *   **Explanation:** After parsing, `BeautifulSoup` allows you to access tags as attributes (e.g., `soup.title`). `.string` then gets the text inside that tag. This works for tags that only contain text.

3.  **`soup.find(name, attrs={}, recursive=True, text=None, **kwargs)`**
    *   **Purpose:** Finds the *first* tag that matches the given criteria.
    *   **Example Usage:**
        ```python
        # Assuming 'soup' is from the previous BeautifulSoup example
        first_h1 = soup.find('h1')
        if first_h1:
            print(f"First H1 tag text: {first_h1.text}")

        # Find a div with a specific class
        main_div = soup.find('div', class_='main-content') # Note 'class_' for the class attribute
        if main_div:
            print(f"Found div with class 'main-content'.")

        # Find an element by ID
        element_by_id = soup.find(id='uniqueId')
        if element_by_id:
            print(f"Found element with ID 'uniqueId'.")
        ```
    *   **Explanation:**
        *   `name`: A filter for tag name (e.g., `'h1'`, `'p'`, `'a'`).
        *   `attrs`: A dictionary of attributes (e.g., `{'class': 'some-class'}`). Note that `class` is a Python keyword, so for a `class` attribute, you use `class_`.
        *   `**kwargs`: You can also directly pass attributes like `id='my_id'` or `class_='node__title'`.

4.  **`soup.find_all(name, attrs={}, recursive=True, text=None, limit=None, **kwargs)`**
    *   **Purpose:** Finds *all* tags that match the given criteria and returns them as a list.
    *   **Example Usage:**
        ```python
        # Assuming 'soup' is from the previous BeautifulSoup example
        all_paragraphs = soup.find_all('p')
        print(f"Found {len(all_paragraphs)} paragraph tags.")
        if all_paragraphs:
            print(f"First paragraph: {all_paragraphs[0].text}")

        # Find all list items with a specific class, limit to 2 results
        list_items = soup.find_all('li', class_='item-list', limit=2)
        if list_items:
            print(f"Found {len(list_items)} list items with class 'item-list'.")
        ```
    *   **Explanation:** Similar to `find()`, but returns a list of all matches. `limit` can restrict the number of results.

5.  **`soup.select(selector)`**
    *   **Purpose:** Finds all elements matching a CSS selector, returning a list of `Tag` objects.
    *   **Example Usage:**
        ```python
        # Assuming 'soup' is from the previous BeautifulSoup example
        # Select all anchor tags within a div with id 'nav'
        nav_links = soup.select('#nav a')
        if nav_links:
            print(f"Found {len(nav_links)} links in nav.")
            for link in nav_links:
                print(f"- {link.text.strip()}")

        # Select all elements with class 'highlight'
        highlighted_elements = soup.select('.highlight')
        if highlighted_elements:
            print(f"Found {len(highlighted_elements)} highlighted elements.")
        ```
    *   **Explanation:** This method allows you to use standard CSS selectors (like you would in JavaScript or browser developer tools) to find elements. It's often more concise and powerful for complex selections than `find_all()`.
        *   `.class`: Selects elements with a specific class.
        *   `#id`: Selects an element with a specific ID.
        *   `tag`: Selects all elements of a specific tag type.
        *   `tag.class`: Selects elements of a specific tag with a specific class.
        *   `tag > child_tag`: Selects direct children.
        *   `tag descendant_tag`: Selects descendants at any level.

6.  **`tag.text`** (or `tag.get_text()`)
    *   **Purpose:** An attribute (or method) on a `Tag` object that returns all the text content within that tag, stripping out any nested HTML tags.
    *   **Example Usage:**
        ```python
        # Assuming 'link_tag' is a BeautifulSoup Tag object, e.g., from soup.find('a')
        link_text = link_tag.text
        print(f"Raw text of link: '{link_text}'")

        # Using get_text() for more options, e.g., to join text with spaces
        # text_with_spaces = some_tag.get_text(separator=' ', strip=True)
        ```
    *   **Explanation:** Useful for extracting just the readable text from an HTML element.

7.  **`string.strip()`**
    *   **Purpose:** A built-in Python string method that removes leading and trailing whitespace (spaces, newlines, tabs) from a string.
    *   **Example Usage:**
        ```python
        dirty_string = "  \n  Some text with spaces  \t\n  "
        clean_string = dirty_string.strip()
        print(f"Original: '{dirty_string}'")
        print(f"Stripped: '{clean_string}'") # Output: 'Some text with spaces'
        ```
    *   **Explanation:** Essential for cleaning up extracted text that often contains extra whitespace due to HTML formatting.

8.  **`tag.get('attribute_name')`**
    *   **Purpose:** A method on a `Tag` object that retrieves the value of a specified HTML attribute.
    *   **Example Usage:**
        ```python
        # Assuming 'link_tag' is a BeautifulSoup Tag object, e.g., <a>
        # <a href="/path/to/page.html" target="_blank">Click Me</a>
        href_value = link_tag.get('href')
        target_value = link_tag.get('target')
        non_existent_attr = link_tag.get('data-foo')

        print(f"href attribute: {href_value}")
        print(f"target attribute: {target_value}")
        print(f"Non-existent attribute 'data-foo': {non_existent_attr}") # Will be None
        ```
    *   **Explanation:** For an `<a href="...">` tag, `link_tag.get('href')` will return the URL. It safely returns `None` if the attribute doesn't exist, avoiding errors.

## 4. Full Script: Scraping a Cybersecurity News Website

Here's a full script to scrape article titles and links from `Dark Reading`, a popular cybersecurity news website. Please note that website structures can change, and aggressive scraping can lead to IP blocking. Always respect `robots.txt` and the website's terms of service.

````python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin


# Website
url = "https://thehackernews.com/"


# Browser-like header
headers = {
    "User-Agent": "Mozilla/5.0"
}


# Send request
response = requests.get(url,headers=headers,timeout=10)


# Check request
# print("Status Code:", response.status_code)

# response.raise_for_status()


# Parse HTML
soup = BeautifulSoup(response.text,"html.parser")


# Find all links
links = soup.find_all("a", href=True)


print("\nThe Hacker News Articles")
print("=" * 80)


count = 0


for link in links:

    title = link.get_text(" ", strip=True)
    href = link.get("href")


    # Ignore empty text
    if not title:
        continue


    # Convert relative URL to absolute URL
    article_url = urljoin(url, href)


    # Make sure it is a THN URL
    if "thehackernews.com" not in article_url:
        continue


    # Ignore very short navigation text
    if len(title) < 30:
        continue


    count += 1


    print(f"\n{count}. {title}")
    print(f"URL: {article_url}")


    # Stop after 10 articles
    if count == 10:
        break
````
