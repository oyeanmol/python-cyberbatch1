# Python schedule Module

## 1. What is the schedule Module?

The schedule module is a Python library used to run functions automatically at specific intervals or times.

For example:
* Run a function every 5 seconds
* Run a function every 10 minutes
* Run a function every hour
* Run a function every day
* Run a function every Monday
* Run a function every day at a specific time

````python
import schedule
````

## 2. Basic Structure

The basic structure of a scheduled task is:

```python
import schedule

def task():
    print("Task executed")

schedule.every(5).seconds.do(task)

while True:
    schedule.run_pending()
```

Here:
* `schedule.every(5).seconds` means: Run the task every 5 seconds.
* `.do(task)` means: Execute the task function.

## 15. schedule.run_all()

`run_all()` runs all scheduled jobs immediately.

Normally `schedule.run_pending()` waits for the scheduled time. But `schedule.run_all()` runs the jobs immediately.

### Example 1

````python
import schedule

def task1():
    print("Task 1 executed")

def task2():
    print("Task 2 executed")

schedule.clear()
schedule.every(10).seconds.do(task1)
schedule.every(20).seconds.do(task2)

schedule.run_all()
````

Output:
```
Task 1 executed
Task 2 executed
```

## 16. schedule.cancel_job()

Sometimes we want to stop a particular scheduled job. For this, we can store the job in a variable:
`job = schedule.every(5).seconds.do(task)`

Then cancel it:
`schedule.cancel_job(job)`

### Example 1

````python
import schedule

def task():
    print("Task running")

schedule.clear()
job = schedule.every(5).seconds.do(task)

schedule.cancel_job(job)
````

The task is cancelled.

### Example 2

````python
import schedule

def task():
    print("Hello")

schedule.clear()
job = schedule.every(5).seconds.do(task)

schedule.cancel_job(job)

schedule.run_all()
````

The task will not execute because it was cancelled.

### Example 3

````python
import schedule

def task1():
    print("Task 1")

def task2():
    print("Task 2")

schedule.clear()
job1 = schedule.every(5).seconds.do(task1)
job2 = schedule.every(10).seconds.do(task2)

schedule.cancel_job(job1)
````

Only task1 is cancelled.

Create a simple Task Scheduler that automatically runs different tasks at different intervals.

### Project Code

However, there is one important issue with this particular approach: `input()` blocks the scheduler loop while waiting for user input. For a beginner project, it's better to keep the scheduler itself simple.

### Recommended Project Version

````python
import schedule
import time

def check_tasks():
    print("Checking tasks")

def generate_report():
    print("Generating report")

def send_reminder():
    print("Sending reminder")

schedule.clear()
schedule.every(5).seconds.do(check_tasks)
schedule.every(10).seconds.do(generate_report)
schedule.every(15).seconds.do(send_reminder)

print("Task Scheduler Started")

# Running loop for 16 seconds to observe non-blocking scheduler output
start_time = time.time()
while time.time() - start_time < 16:
    schedule.run_pending()
    time.sleep(1)
````

### Most Important Methods

| Method | Purpose |
| :--- | :--- |
| `schedule.every()` | Create a scheduled job |
| `.seconds` | Schedule in seconds |
| `.minutes` | Schedule in minutes |
| `.hours` | Schedule in hours |
| `.days` | Schedule in days |
| `.weeks` | Schedule in weeks |
| `.monday` etc. | Schedule on a specific weekday |
| `.at()` | Run at a specific time |
| `.do()` | Attach a function to the job |
| `run_pending()` | Execute jobs that are due |
| `run_all()` | Execute all jobs immediately |
| `cancel_job()` | Cancel a particular job |
| `clear()` | Remove scheduled jobs |

**Core concept:** `schedule` does not execute your function by itself just because you created a schedule. You normally keep calling `schedule.run_pending()` so the scheduler can check which jobs are due.
