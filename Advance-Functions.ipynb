```python
# Main simulation logic
users_database = {}

def welcome():
    print("-- Welcome to Hotel --\n")
    print("1. User Registration")
    print("2. User Login")
    print("3. Exit")
def register_user(username, password, users_db):
    """Registers a new user if the username is not already taken."""
    if username in users_db:
        return False, "Username already exists."
    else:
        users_db[username] = password
        return True, "Registration successful."

def login(username, password, user_db1):
    if username in user_db1 and user_db1[username] == password:
        return True, print("Login Successful")
    elif username not in user_db1:
        print("User not found! Error")
    else:
        return False, print("Incorrect Password")
def login_user(username, password, users_db):
    """Authenticates a user against the users database."""
    if username in users_db and users_db[username] == password:
        return True, "Login successful!"
    elif username not in users_db:
        return False, "Username not found."
    else:
        return False, "Incorrect password."

while True:
    welcome()
    choice = input("Enter your choice: ")

    if choice == '1':
        uname = input("Enter desired username: ")
        pwd = input("Enter desired password: ")
        success, message = register_user(uname, pwd, users_database)
        print(message)
        if success:
            print("Current users:", users_database)
    elif choice == '2':
        uname = input("Enter username: ")
        pwd = input("Enter password: ")
        success, message = login_user(uname, pwd, users_database)
        print(message)
    elif choice == '3':
        print("Exiting program. Goodbye!")
        break
    else:
        print("Invalid choice. Please try again.")
```
